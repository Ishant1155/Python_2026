Given a list of integers, extract even numbers, calculate their squares, and then extract squares greater than 100.


```python
a = [1, 2, 12, 21, 7]
print("Original list",a)
even = list(filter(lambda x: x % 2 == 0, a))
print("even numbers:",even)
square=list(map(lambda x:x**2,a))
extract=list(filter(lambda x:x>100,square))
print(extract)
```

Given a list of strings, extract strings with length greater than 3 and convert them into uppercase.

```python
str=["hello","hi","ola","namaste"]
print(str)
greater=list(filter(lambda x:len(x)>3,str))
caps=list(map(lambda x: x.upper(),greater))
print(caps)
```

Write a program to extract numbers divisible by 2 or 3 and calculate their squares.

```python
num=[2,3,5,6,7,9]
print(num)
extract=list(filter(lambda x:x%2==0 or x%3==0,num))
square=list(map(lambda x:x**2,extract))
print(extract)
print(square)
```

Given a list of integers, add 10 to every element and then extract elements greater than 30.

```python
num=[1,6,12,19,21,23]
print(num)
add=list(map(lambda x:x+10,num))
print(add)
```

Write a program to extract negative numbers and convert them into their absolute values.

```python
num=[-21,-10,-23,2,4]
absolute=list(map(lambda x: abs(x),num))
print(absolute)
```

Given a list of words, extract words ending with the letter 'n' and calculate their lengths.

```python
words=["new","police","pan","negation","knight"]
with_n=list(filter(lambda x:x[-1]=="n",words))
print(with_n)
```
