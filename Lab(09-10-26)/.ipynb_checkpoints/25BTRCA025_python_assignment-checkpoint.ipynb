Accept a paragraph, convert words to lowercase, remove basic punctuation (commas, periods and exclamation marks), create a set of unique words, count total and unique words, identify repeated words, and display unique words alphabetically. Do not use external libraries.

```python
paragraph = input("Enter a paragraph: ")
paragraph = paragraph.lower()
paragraph = paragraph.replace(",", "")
paragraph = paragraph.replace(".", "")
paragraph = paragraph.replace("!", "")
words = paragraph.split()
unique_words = set(words)
total_words = len(words)
unique_count = len(unique_words)
repeated_words = []
for word in unique_words:
    if words.count(word) > 1:
        repeated_words.append(word)
sorted_words = sorted(unique_words)
print("\nTotal number of words:", total_words)
print("Number of unique words:", unique_count)

print("\nRepeated words:")
print(repeated_words)

print("\nUnique words in alphabetical order:")
print(sorted_words)
```
