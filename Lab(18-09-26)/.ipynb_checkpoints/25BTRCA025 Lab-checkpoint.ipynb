Create a list of 10 numbers and print only even numbers.

```python

numbers = [10, 15, 20, 25, 30, 35, 40, 45, 50, 55]

print("Even numbers are:")

for num in numbers:
    if num % 2 == 0:
        print(num)
```

Find the second-largest value in a list.

```python
numbers = [10, 25, 45, 30, 50, 15]

numbers = list(set(numbers))
numbers.sort()

print("Second largest number:", numbers[-2])
```

Create a tuple of 5 subjects and unpack it.

```python
subjects = ("Python", "Design Thinking", "Maths", "Operating System", "Biology")

a, b, c, d, e = subjects

print(a)
print(b)
print(c)
print(d)
print(e)
```

Remove duplicates from a list using a set.

```python
numbers = [10, 20, 30, 20, 40, 10, 50, 30]

unique_numbers = list(set(numbers))

print("Original list:", numbers)
print("List without duplicates:", unique_numbers)
```

Create a dictionary of 5 students and their marks.

```python
students = {
    "Ishant": 90,
    "Tribendra": 85,
    "Hridaya": 20,
    "Riya": 95,
    "Priya": 88
}

print("Student Marks:")

for name, marks in students.items():
    print(name, ":", marks)
```

Count character frequencies in a string using a dictionary.

```python
string = "programming"

frequency = {}

for char in string:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print("Character frequencies:")

for char, count in frequency.items():
    print(char, ":", count)
```
