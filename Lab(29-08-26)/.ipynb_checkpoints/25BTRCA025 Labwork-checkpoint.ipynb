Fibonacci series

```python
n = 5
a, b = 0, 1
for i in range(n + 1):
    print(a, end=" ")
    a, b = b, a + b
```

Multiplication table

```python
n = int(input("Enter a number for tables: "))
m=int(input("Enter the limit: "))
print("Multiplication table is", n, "is: ")
for i in range(1, m+1):
    print(n, "X", i, "=", n * i)
```

Triangle star loop

```python
n = 5
for i in range(1, n + 1):
    space = n - i
    star = 2 * i - 1
    print(" " * space + "*" * star)
```

Right hand star loop triangle

```python
n = 5
for i in range(n):
    for j in range(i + 1):
        print("*", end=" ")
    print()
```

```python
Inverted star loop triangle 
```

```python
n = 5
for i in range(n):
    for j in range(i, n):
        print("*", end=" ")
    print()
```
